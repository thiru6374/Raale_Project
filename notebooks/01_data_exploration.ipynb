# 01 - Data Exploration
Exploratory analysis of the synthetic Chennai neighbourhood dataset.

```python
import sys
sys.path.insert(0, '..')

from src.data.ingestion import DataIngestor
from src.data.preprocessing import DataPreprocessor
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate synthetic data (5% missing rate)
valid_df, errors = DataIngestor.load_synthetic_data(num_records=200, missing_rate=0.05)
print(f'Valid records: {len(valid_df)}, Rejected: {len(errors)}')
valid_df.head()
```

## Basic Statistics

```python
valid_df.describe()
```

## Missing Values

```python
missing = valid_df.isnull().sum()
missing[missing > 0]
```

## Temperature Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
valid_df['temperature_c'].hist(bins=30, ax=axes[0], color='#FF6B6B', edgecolor='black')
axes[0].set_title('Temperature Distribution (Chennai)')
axes[0].set_xlabel('Temperature (°C)')
valid_df['heat_index'].hist(bins=30, ax=axes[1], color='#FF8E53', edgecolor='black')
axes[1].set_title('Heat Index Distribution')
axes[1].set_xlabel('Heat Index (°C)')
plt.tight_layout()
plt.savefig('../data/processed/temperature_distribution.png', dpi=150)
plt.show()
```

## Vulnerability vs Built Density Correlation

```python
plt.figure(figsize=(8, 6))
scatter = plt.scatter(valid_df['built_density'], valid_df['vulnerability_index'],
                      c=valid_df['temperature_c'], cmap='Reds', alpha=0.7, s=60)
plt.colorbar(scatter, label='Temperature (°C)')
plt.xlabel('Built Density')
plt.ylabel('Vulnerability Index')
plt.title('Built Density vs Vulnerability (coloured by Temperature)')
plt.savefig('../data/processed/density_vs_vulnerability.png', dpi=150)
plt.show()
```

## District Distribution

```python
valid_df['district'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Neighbourhood count by District')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Key Observations

- Temperature ranges roughly from 32°C to 41°C in the simulation (pre Urban Heat Island effect).
- Heat Index pushes some zones into the 40–45°C range — matching Chennai's real summer conditions.
- Built density strongly correlates with green cover deficits.
- Vulnerable populations (elderly + low income) are distributed across districts.
