# Phase 5: Experiment & Evaluation Notebook

This notebook demonstrates how to run the Phase 5 evaluation architecture to compare outreach optimization strategies.

```python
import os
import sys

# Ensure project root is in python path
sys.path.append(os.path.abspath('..'))

import pandas as pd
from src.evaluation.experiment_runner import ExperimentRunner
from src.evaluation.evidence_generator import EvidenceGenerator
from src.evaluation.failure_analysis import FailureAnalyzer
```

## 1. Run Optimization Strategies

We compare COVERAGE_FOCUSED against FAIRNESS_AWARE on the exact same synthesized dataset.

```python
runner = ExperimentRunner(num_records=150, missing_rate=0.05)
results = runner.run_comparison()

print(f"Experiment Status: {results['status']}")
```

## 2. Compare Metrics

```python
cov = results['coverage_focused']['metrics']
fair = results['fairness_aware']['metrics']
base = cov.get('baseline_comparison', {})

df_metrics = pd.DataFrame([
    {
        'Strategy': 'Baseline',
        'High Risk Coverage %': base.get('baseline_high_risk_coverage_pct', 0)
    },
    {
        'Strategy': 'Coverage-Focused',
        'High Risk Coverage %': cov['high_risk_coverage_pct']
    },
    {
        'Strategy': 'Fairness-Aware',
        'High Risk Coverage %': fair['high_risk_coverage_pct']
    }
])

display(df_metrics)
```

## 3. Generate Defensible Evidence

```python
generator = EvidenceGenerator()
report_path = generator.generate_report(results)
print(f"Evidence report saved to: {report_path}")
```

## 4. Run Failure Mode Analysis

Testing how the system behaves under low capacity.

```python
analyzer = FailureAnalyzer(num_records=50)
failure_result = analyzer.run_scenario("LOW_CAPACITY")
print(f"Scenario Status: {failure_result['status']}")
print(f"Capacity Utilization: {failure_result['metrics']['capacity_utilization_pct']}%")
```
